```python
import os
from pathlib import Path
```

```python
base = Path.cwd().parents[1]

folder1_up = base / "R_code&data" / "results" / "primary" / "ensembl" / "up"
folder2_up = base / "R_code&data"/"results_on_sample_var1"/"primary"/"ensembl"/"up"

folder1_down = base /"R_code&data"/"results"/"primary"/"ensembl"/"down"
folder2_down = base /"R_code&data"/"results_on_sample_var1"/"primary"/"ensembl"/"down"
```

```python
#functions to format strings and to match file names
def read_strings(path):
    with open(path, "r", encoding="utf-8") as f:
        result = set()
        for line in f:
            line = line.strip()
            if not line:
                continue

            # Remove the start number if it is
            parts = line.split(maxsplit=1)
            if len(parts) == 2 and parts[0].isdigit():
                line = parts[1]

            result.add(line)

        return result


def find_matching_file(filename, files):
    name = os.path.splitext(filename)[0].lower()

    matches = []
    for f in files:
        other = os.path.splitext(f)[0].lower()

        if name in other or other in name:
            matches.append(f)

    return matches
```

```python
#list the ensembl that appear in individual cases up-regulated genes (custom variant vs standard variant)
files1 = os.listdir(folder1_up)
files2 = os.listdir(folder2_up)

all_only_up = []

for f1 in files1:
    if not f1.endswith(".txt"):
        continue

    matches = find_matching_file(f1, files2)

    if len(matches) == 0:
        print(f"Nu exista pereche pentru {f1}")
        continue

    if len(matches) > 1:
        print(f"Perechi multiple pentru {f1}: {matches}")
        continue

    f2 = matches[0]

    set1 = read_strings(os.path.join(folder1_up, f1))
    set2 = read_strings(os.path.join(folder2_up, f2))

    only1 = set1 - set2
    only2 = set2 - set1



    if only1:
        all_only_up.extend(x for x in only1 if x not in all_only_down)
        print("\n")
        print(f"{f1}")

        for s in sorted(only1):
            print("  ", s)
```

```python
#list the ensembl that appear in individual cases down-regulated genes (custom variant vs standard variant)
files1 = os.listdir(folder1_down)
files2 = os.listdir(folder2_down)

all_only_down = []

for f1 in files1:
    if not f1.endswith(".txt"):
        continue

    matches = find_matching_file(f1, files2)

    if len(matches) == 0:
        print(f"Nu exista pereche pentru {f1}")
        continue

    if len(matches) > 1:
        print(f"Perechi multiple pentru {f1}: {matches}")
        continue

    f2 = matches[0]

    set1 = read_strings(os.path.join(folder1_down, f1))
    set2 = read_strings(os.path.join(folder2_down, f2))

    only1 = set1 - set2
    only2 = set2 - set1

    if only1:
        all_only_down.extend(x for x in only1 if x not in all_only_down)
        print("\n")
        print(f"{f1}")

        for s in sorted(only1):
            print("  ", s)

```

```python
#the number of all up or down regulated genes found only in the custom variant
print(len(all_only_up))
print(len(all_only_down))
```

```python
#transform the codes of down-regulated genes in hugo codes

import requests
server = "https://rest.ensembl.org"
endpoint = "/lookup/id"

r = requests.post(
    server + endpoint,
    headers={"Content-Type": "application/json",
             "Accept": "application/json"},
    json={"ids": all_only_down}
)

data = r.json()

for ens, info in data.items():
    print(ens, "->", info.get("display_name"))
```

```python
#see how many hogo codes are
hugos_down=[]
for ens, info in data.items():
    if info.get("display_name"):
        hugos_down.append(info.get("display_name"))

len(hugos_down)
```
