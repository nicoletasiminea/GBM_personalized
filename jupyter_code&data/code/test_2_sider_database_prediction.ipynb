```python
import pandas as pd
```

```python
#make a table with meddra results and a list with all the drugs obtained in our results, then extract all the reactions for the drug list and save them
df_drug_names = pd.read_csv("../data/drug_names.tsv", sep="\t", header=None, names=["CID", "Drug"])
df_meddra = pd.read_csv("../data/meddra_all_se.tsv", sep="\t", header=None, usecols=[0, 5],names=["CID","Reaction"])
resultat = df_drug_names.merge(df_meddra, on="CID")

df = pd.read_excel("../data/drugs_for_each_case_in_one_sheet.xlsx")
drugs_list=df["Drug"].unique()
drug_list = [x.lower() for x in drugs_list]


reactions = []

for x in drug_list:
    val = (
        resultat.loc[resultat["Drug"] == x, "Reaction"]
          .dropna()
          .unique()
          .tolist()
    )

    reactions.append({
        "Drug": x,
        "Reaction": ", ".join(map(str, val))
    })


reactions = pd.DataFrame(reactions)
reactions = reactions[reactions["Reaction"] != ""].reset_index(drop=True)
reactions.to_excel("../data/Sider_reactions.xlsx")
```
